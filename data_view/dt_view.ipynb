```python
import pandas as pd 
import sqlalchemy

db_url = "postgresql://postgres:viacep_etl@localhost:5432/db_etl_viacep"
db_path = db_url

with open("../data/03-gold_enriched/users_cep_info.sql", 'r', encoding='utf-8') as f:
    query = f.read()

with sqlalchemy.create_engine(db_path).connect() as conn:
    df = pd.read_sql_query(query, conn)

df.head()
```
